```text


**STRUTTURA DEL PROGETTO SUL REPOSITORY MLOps** <br>
```text
project/
¦
+-- app/
¦   +-- main.py
¦   +-- routes/
¦   +-- services/
¦   +-- utils/
¦
+-- tests/
¦   +-- CI/
¦   +-- CD/
¦
+-- model/
¦   +-- model.pkl
¦
+-- airflow/
+-- monitoring/
¦
+-- README.md
+-- docs/
¦   +-- architecture.md
¦   +-- ml_pipeline.md
¦   +-- monitoring.md
¦   +-- deployment.md
¦   +-- api_reference.md
¦
+-- architecture/
+-- notebooks/
+-- diagrams/
¦
+-- links:
¦   +-- Hugging Face
¦   +-- Grafana screenshots
¦   -- Colab demo
¦
+-- requirements.txt
+-- Dockerfile
+-- setup.sh


**SNIPPET DI CODICE ED ESEMPI:MLFLOW** <br>
Dettaglio del codice per fare Logging delle metriche Durante il training.<br>
Vedere anche la demo del modello nei notebook dimostrati indicati nella sezione dei link pubblici. <br>
[Esempio training sentiment analisys con RoBERta](https://colab.research.google.com/drive/15vppELulgC2xW97ivuhQNWVM913_kSIh?usp=sharing)
```text


trainer = Trainer( model=model,
                  args=training_args,
                  train_dataset=tokenized_dataset["train"],
                  eval_dataset=tokenized_dataset["validation"],
                  compute_metrics=compute_metrics )

trainer.train()
# Una volta finito, salva manualmente il tokenizer nella cartella dei risultati
tokenizer.save_pretrained("./sentiment-model-base")


# Nome esperimento MLflow
mlflow.set_experiment("twitter_sentiment_roberta_base")

# 1. IMPOSTA PRIMA IL TRACKING URI
mlflow.set_tracking_uri("file:./mlruns")

# 2. DEFINISCI IL NOME DELL'ESPERIMENTO
experiment_name = "twitter_sentiment_roberta_base"

# 3. RECUPERA O CREA L'ESPERIMENTO IN MODO SICURO
experiment = mlflow.get_experiment_by_name(experiment_name)
if experiment is None:
    experiment_id = mlflow.create_experiment(experiment_name)
else:
    experiment_id = experiment.experiment_id

# 4. IMPOSTA L'ESPERIMENTO ATTIVO TRAMITE IL SUO ID REALE
mlflow.set_experiment(experiment_id=experiment_id)


with mlflow.start_run(run_name="twitter-sentiment-roberta"):

    mlflow.set_tag("project", "twitter_sentiment")
    mlflow.set_tag("framework", "huggingface")
    mlflow.set_tag("model_family", "roberta")

    # log parametri custom   
    mlflow.log_param("model_name", MODEL_NAME)
    mlflow.log_param("dataset", "tweet_eval")
    mlflow.log_param("max_length", 128)

    trainer.train()

    # valutazione finale
    metrics = trainer.evaluate()

    # log metriche finali
    mlflow.log_metrics(metrics)

    # salva modello HuggingFace
    mlflow.transformers.log_model(transformers_model={
                                                    "model": trainer.model,
                                                    "tokenizer": tokenizer
                                                        },
                                                    artifact_path="sentiment_model-base"
                                 )

**SNIPPET DI CODICE ED ESEMPI:EXPORT DATI CON CSV / SQLite** <br>
```text
# CSV con pandas
metrics = {
    "accuracy": acc,
    "f1": f1,
    "timestamp": datetime.now()
}
df = pd.DataFrame([metrics])
df.to_csv("metrics.csv", mode="a", index=False)

# SQLlite non  usa CSV ma usa direttamente un database SQLite: metrics.db
# SQLite è semplicemente: metrics.db cioè un file dentro il quale c’è tutto il # database. Se il file NON esiste: SQLite lo crea automaticamente.

# La soluzione SQLite è più professionale di CSV,non richiede setup e simula un # DB reale. Grafana legge le tabelle del DB e s'integra bene con Airflow e python
      
# Creazione del db
import sqlite3
conn = sqlite3.connect("metrics.db")
# End Creazione del db se non esiste crea il file

# Start Creazione una tabella model_metrics_baseline
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS model_metrics_baseline (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    timestamp TEXT,
    accuracy REAL,
    f1_score REAL
)
""")

conn.commit()
conn.close()
# End Creazione tabella model_metrics_baseline

# Start Creazione tabella model_metrics_finetuning
# se la tabella non esiste viene creata
df.to_sql(
    "model_metrics_finetuning",
    conn,
    if_exists="append",
    index=False
)
# End Creazione una tabella model_metrics_finetuning

Come risultato abbiamo sempre una tabella metrics
| timestamp | accuracy | f1_score |
| --------- | -------- | -------- |
| ...       | ...      | ...      |

**Analisi e strategie per mostrare i grafici GRAFANA agli utenti** <br>
Riepilogo dei 3 Ambienti (per non confondersi)
Il Repository (GitHub.com): <br>
"hard disk" remoto che Contiene i file di testo, il file SQLite e il README. È statico, memorizza solo i dati. <br>

GitHub Codespace: ambiente di lavoro interattivo. <br>
usato per programmare ed eseguire il training. Se si spegne, non succede nulla ai file salvati
(committati) nel Repository. <br>

La GitHub Action: <br>
È il tuo "assistente robotico" nel cloud. Si sveglia quando aggiorni il file SQLite nel repository,
accende un server, esegue install_db.sh, avvia Docker, scatta la foto al grafico di Grafana, la salva nel repository e poi scompare. <br>

i GitHub Codespaces nascono come ambienti di sviluppo temporanei e non sono pensati per fare da server di produzione
(o runtime) sempre attivi. Dopo qualche minuto di inattività (di default 30 minuti),
il Codespace va in timeout e si spegne completamente, interrompendo anche il tuo Docker Compose. <br>

editare il file docker-compose.yaml con le impostazioni di grafana
suggerite ed eseguire

docker compose up -d grafana
Monitoring Reference Guide on https://grafana.com/docs/grafana/latest/setup-grafana/installation/docker/ <br>

Entrare la prima volta come admin + admin <br>
Creare un nuovo utente con i seguenti valori: <br>
nome: Francesco Frigerio <br>
mail: francesco.frigerio71@gmail.com <br>
role: admin  <br>
pwd: stessa pwd definita nella creazione(vedere setup) <br>


```text
# check docker compose installation
docker compose version
# first go into the directory where you have created this docker-compose.yaml file
cd monitoring

# create the docker-compose.yaml file
touch docker-compose.yaml

# edit yaml file
services:
  grafana:
    image: grafana/grafana-enterprise
    container_name: grafana
    restart: unless-stopped
    ports:
      - '3000:3000'
    volumes:
      - grafana-storage:/var/lib/grafana
# Use Docker volumes when you want the Docker Engine to manage the storage volume.
# To use Docker volumes for persistent storage,
volumes:
  grafana-storage: {}

# start/stop the grafana container
# d = detached mode
# up = to bring the container up and running
# To determine that Grafana is running,
# open a browser window and type IP_ADDRESS:3000. The sign in screen should appear.
docker compose up -d
# down = to bring the container down and stopping
docker compose down

**INSTALLAZIONE SQLITE SU GRAFANA** <br>

installazione plugin sqlite
Dall'interfaccia grafica (UI):
Vai su Administration > Plugins and data > Plugins, cerca "SQLite" e clicca su Install.

Tramite riga di comando (CLI):
con accesso al server eseguire questo comando:
grafana-cli plugins install frser-sqlite-datasource

(riavviare il servizio Grafana dopo l'installazione tramite CLI, ad esempio con sudo systemctl restart grafana-server)*.

---

### 2. Configurazione della sorgente dati
Una volta installato il plugin:
1. Vai su **Connections** > **Data Sources** e clicca su **Add data source**.
2. Cerca **SQLite** e selezionalo.
3. Nel campo **Path**, dovrai inserire il percorso assoluto in cui si trova il file del tuo database (es. /opt/machineinnovation/db/metrics.db).

---


* **I permessi del file metrics.db:** Il file .db di SQLite deve essere tassativamente leggibile dall'utente di sistema
che esegue Grafana (di solito l'utente grafana). Se il file si trova nella cartella home di un altro utente, Grafana
restituirà un errore di *"Permission Denied"*.
* **La sicurezza e systemd:** Se usi Linux e hai installato Grafana tramite gestore pacchetti,
il servizio systemd di Grafana blocca di default l'accesso alla cartella /home. Se il tuo database si trova lì,
la soluzione migliore è spostarlo in una directory comune (come /var/lib/grafana/ o /opt/)
e assegnargli i giusti permessi. <br>

**QUERY SU GRAFANA** <br>

SELECT
  -- 1. Converte il timestamp nel formato
  -- richiesto da Grafana (Unix Epoch in secondi)
  CAST(strftime('%s', timestamp) AS INTEGER) AS time,
  
  -- 2. Seleziona le metriche da visualizzare sul grafico
  accuracy AS "Accuracy",
  precision AS "Precision",
  recall AS "Recall",
  f1_score AS "F1-Score"

FROM model_metrics_baseline

-- 3. Applica il filtro temporale nativo di Grafana
-- Divide per 1000 perché Grafana ragiona in millisecondi,
-- mentre SQLite (e strftime %s) in secondi
WHERE CAST(strftime('%s', timestamp) AS INTEGER) >= $__from / 1000
  AND CAST(strftime('%s', timestamp) AS INTEGER) < $__to / 1000

-- 4. Ordina i dati dal più vecchio al più recente per permettere a Grafana di tirare le linee correttamente
ORDER BY timestamp ASC








**MONITORING CON GRAFANA E CODESPACE ATTIVO** <br>
1. mostrare i grafici (Con Codespace attivo) <br>
Biosgna seguire docker compose up alla partenza del Codespace
Serve solo Se si decide che l'utente debba usare il Codespace per vedere i grafici, automatizzare
l'avvio è la scelta migliore per evitare che debba digitare comandi nel terminale.

Per farlo configurare il file .devcontainer/devcontainer.json nel repository. Aggiungendo la proprietà postStartCommand,
il Codespace avvierà automaticamente Grafana ogni volta che viene riattivato.
Anche automatizzando l'avvio,
l'URL generato da GitHub Codespaces (es. https://random-space-3000.app.github.dev) è temporaneo e scade.
 Fornendo quell'URL statico nel file README.md, il giorno dopo non funzionerà più.
 Per questo motivo è preferibile la opzione 1.2

```text
JSON
{
  "name": "Mio Training Env",
  "dockerComposeFile": "docker-compose.yml",
  "service": "tuo-servizio-principale",
  "postStartCommand": "docker compose up -d grafana",
  "portsAttributes": {
    "3000": {
      "label": "Grafana Dashboard",
      "onAutoForward": "openPreview" // Apre automaticamente i grafici all'avvio
    }
  }
}


**MONTITORING CON GRAFANA E CODESPACE NON ATTIVO** <br>

2. L'Alternativa per mostrare i grafici (Senza Codespace attivo)
Poiché il Codespace va spesso in down, l'idea di fornire immagini statiche o report è eccellente e molto più affidabile
per un repository pubblico. Ecco le opzioni migliori:

Opzione 2.1 : GitHub Actions + Grafana Image Renderer (La più elegante)
Puoi fare in modo che ogni volta che il tuo training finisce e aggiorna il file SQLite,
una GitHub Action si attivi automaticamente, avvii Grafana in background, catturi lo screenshot dei grafici
e salvi l'immagine (es. dashboard.png) direttamente nel repository. In questo modo, l'immagine nel tuo README.md sarà sempre aggiornata all'ultimo training.

Come funziona: La pipeline usa lo strumento ufficiale grafana-image-renderer.

Vantaggio: Gli utenti vedono i grafici istantaneamente nel README.md senza dover avviare o cliccare nulla.

Opzione 2.2 : Esportare la Dashboard in PDF / Snapshot
Se si preferisce non automatizzare tutto, si può generare manualmente dei report da allegare alle Releases del tuo repository
o alla cartella dei risultati:

Share Snapshot: Grafana permette di creare uno "Snapshot" (una copia statica interattiva dei dati).
Puoi esportarla come file JSON o caricarla sul cloud pubblico di Grafana (snapshot.raintank.in)
e condividere quel link nel README.

Export as PDF/Image: Dalle opzioni della dashboard di Grafana puoi fare clic su Share -> Export as PDF
o Export as image e salvare il file direttamente nel tuo repository GitHub.

Il mix perfetto per un repository GitHub di successo in questo scenario è:
screenshot statico aggiornato (Opzione A o B) direttamente nel file README.md
in modo che l'utente veda subito i risultati del training.

Inserire un pulsante "Open in GitHub Codespaces". Se l'utente vuole esplorare i grafici in modo interattivo
(cambiare i filtri di tempo, fare zoom, ecc.), cliccherà sul pulsante, il Codespace si avvierà,
eseguirà il docker compose up in automatico e gli aprirà Grafana davanti agli occhi.


La soluzione migliore è eseguire lo script install_db.sh
all'interno della GitHub Action (Opzione 2.1).

Eseguire lo script alla fine del training sul codespace,
copieresti il database nella cartella /opt della tua macchina.
 Tuttavia, la GitHub Action gira su un server cloud di GitHub completamente separato e "pulito",
 che non ha accesso ai file del codespace, a meno che tu non faccia il push di quei file nel repository.

Come deve funzionare il flusso dei dati
nel Codespace: Il training finisce e genera il file SQLite nella cartella di output del progetto.

Push su GitHub: Fai il git commit e git push del nuovo file SQLite sul tuo repository pubblico.

Trigger della GitHub Action: GitHub si accorge del push e avvia la pipeline (la GitHub Action).

Esecuzione dello script nel Cloud: All'interno della GitHub Action,
lo script viene eseguito per spostare il file in /opt (del server virtuale di GitHub),
Grafana si avvia, genera l'immagine e la salva nel repository.

Esempio pratico di Workflow GitHub Actions (.github/workflows/update-graphs.yml)
Ecco un esempio di come configurare il file : automatizzare l'esecuzione dello script,
avviare Grafana, fare lo screenshot e aggiornare il repository. <br>

Note
1) Nel README.md:
Ti basterà inserire il tag dell'immagine che punta al file statico,
ad esempio: ![Grafico Training](./images/dashboard.png).
Ogni volta che la GitHub Action finisce, l'immagine cambierà e chiunque
visiti il tuo repository vedrà il grafico aggiornato.

I percorsi nel tuo docker-compose.yml: Assicurati che nel docker-compose
il volume di Grafana punti a /opt del sistema ospitante (es. - /opt:/var/lib/grafana/sqlite_data),
così funzionerà sia nel tuo Codespace/locale sia all'interno dell'ambiente Ubuntu della GitHub Action.

Token di Grafana: Nel passo 5 avrai bisogno di una chiave API di Grafana (Service Account token) c
he puoi generare una volta sola e salvare nei Settings -> Secrets and variables -> Actions del tuo repository GitHub
con il nome GRAFANA_API_KEY. <br>


```text
YAML
name: Aggiorna Grafici README
on:
  push:
    paths:
      - 'percorso/alla/tua/dir_di_output/*.db' # Si attiva solo se cambia il file SQLite

jobs:
  render-charts:
    runs-on: ubuntu-latest
    
    steps:
    # 1. Scarica il codice e tutti i files del tuo repository sul server di GitHub
    - name: Checkout repository
      uses: actions/checkout@v4

    # 2. Configura i permessi ed esegui il tuo script install_db.sh
    # Qui creiamo la cartella /opt sul server di GitHub e vi copiamo il file
    - name: Esegui install_db.sh
      run: |
        chmod +x ./install_db.sh
        # crea la dir sotto /opt se nm esiste
        sudo ./install_db.sh

    # 3. Avvia il tuo Docker Compose (Grafana leggerà il file da /opt del server GitHub)
    - name: Avvia Grafana
      run: docker compose up -d grafana

    # 4. Attendi che Grafana sia pronto (es. 10 secondi)
    - name: Attendi avvio
      run: sleep 10

    # 5. Cattura lo screenshot della dashboard usando Grafana Image Renderer via curl
    # Sostituire l'URL con i parametri corretti della dashboard
    - name: Cattura Screenshot Dashboard
      run: |
        curl -H "Authorization: Bearer ${{ secrets.GRAFANA_API_KEY }}" \
        "http://localhost:3000/render/d-solo/tuo-id-dashboard/nome-dashboard?panelId=1&width=1000&height=500" \
        -o ./images/dashboard.png

    # 6. Spegni i container Docker per liberare risorse
    - name: Ferma Docker
      run: docker compose down

    # 7. Fai il push automatico dell'immagine aggiornata nel tuo repository
    - name: Commit e Push del grafico aggiornato
      run: |
        git config --local user.email "action@github.com"
        git config --local user.name "GitHub Action"
        git add ./images/dashboard.png
        git commit -m "Auto: Aggiornato grafico di training [skip ci]" || exit 0
        git push




**IMPOSTAZIONI GITHUB ACTIONS SU GITHUB**
Vai su GitHub nella pagina del tuo repository. <br>
Clicca su Settings (Impostazioni) in alto a destra. <br>
Nel menu a sinistra, clicca su Actions $\rightarrow$ General.Scorri fino in fondo alla pagina fino alla sezione Workflow permissions.Seleziona l'opzione "Read and write permissions". <br>

**CONCLUSIONE ANALISI  MONITORING** <br>

Abbiamo tre ambienti (Computer locale/Codespace,
Server delle Actions, e il Repository).

Quando usiamo le GitHub Actions, lo script  install_db.sh viene eseguito su una macchina virtuale
"temporanea" messa a disposizione da GitHub nei suoi data center (chiamata in gergo Runner).

Il Codespace è la macchina di sviluppo (dove scriviamo il codice e facciamo partire il training).
La GitHub Action è un server separato che si accende solo quando facciamo git push,
fa il suo lavoro in automatico, e poi si spegne cancellando tutto.

2. essere sul cloud delle Actions non significa lavorare direttamente "dentro" il repository.

Quando la GitHub Action parte, compie questi passaggi:
Si accende un server Ubuntu vuoto nel cloud.
Il comando uses: actions/checkout@v4 fa una copia (un "clone") dei tuoi file dal repository a questo server temporaneo.
Lo script genera la nuova immagine del grafico (dashboard.png) all'interno di questo server temporaneo.
Se la pipeline finisse qui, l'immagine dashboard.png appena creata rimarrebbe dentro quel server e verrebbe cancellata
per sempre un secondo dopo, all'arresto della macchina virtuale.
I comandi Git finali servono proprio a questo:

Solo grazie a questo passaggio l'immagine nel README.md risulterà effettivamente
aggiornata per gli utenti che visitano la pagina del progetto.


**SNIPPET DI CODICE ED ESEMPI:SCHEDULAZIONE CON AIRFLOW** <br>
1. Carica dati <br>
2. Train modello <br>
3. Test modello <br>
4. Log MLflow <br>
5. export delle metriche in SQLLITE <br>
6. Salva metriche DB <br>
7. Aggiorna dashboard <br>





**ANALISI MODEL SERVING** <br>
1. Creare lo Space su Hugging Face
Andare su huggingface.co/spaces e clicca su Create new Space.
Scegliere un nome per il tuo Space.
Fondamentale: Sotto Select the Space SDK, selezionare Docker.
Scegliere il template Blank (vuoto).
Scegliere se farlo pubblico o privato e clicca su Create Space. <br>




**ANALISI MODEL SERVING** <br>
**2. Preparare i file del progetto** <br>


Hugging Face leggerà i file direttamente dalla tua repository dello Space (che funziona esattamente come un repository Git).  <br>
Bisogna creare tre file fondamentali nella root del progetto. <br>
Vedere lezione e pratica del corso <br>
2.1 requirements (fastapi uvicorn) lo creo come in setup.sh (togliere dal setup.sh) <br>
```text
fastapi
uvicorn
transformers
torch
sentencepiece
pydantic

2.2 Sorgente python per implementare l'app Model Serving <br>

2.2) app.py
```text
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from transformers import pipeline
import uvicorn
import os

# 1. Inizializzazione dell'app FastAPI
```text
app = FastAPI(
    title="RoBERTa Sentiment Analysis API",
    description="API per l'analisi del sentiment di un testo utilizzando il modello RoBERTa",
    version="1.0"
)

# 2. Caricamento del modello RoBERTa
```text
# Nota: "cardiffnlp/twitter-xlm-roberta-base-sentiment" funziona molto bene su più lingue, incluso l'italiano.
# Se il tuo testo è SOLO in inglese, puoi usare: "roberta-base" (se finetunato) o "cardiffnlp/twitter-roberta-base-sentiment"
MODEL_NAME = "cardiffnlp/twitter-xlm-roberta-base-sentiment"

print({"status": f"Caricamento del modello {MODEL_NAME} in corso..."})
try:
    # Usiamo la pipeline di Hugging Face per gestire tokenizzazione e inferenza in un colpo solo
    sentiment_pipeline = pipeline("sentiment-analysis", model=MODEL_NAME)
    print({"status": "Modello caricato con successo!"})
except Exception as e:
    print(f"Errore nel caricamento del modello: {e}")
    sentiment_pipeline = None

# 3. Definizione della struttura dei dati in ingresso (Pydantic)
class TextRequest(BaseModel):
    text: str

# Mappatura delle etichette del modello specifico di CardiffNLP per renderle più leggibili
# (Questo modello restituisce di solito: LABEL_0 -> Negative, LABEL_1 -> Neutral, LABEL_2 -> Positive)
LABEL_MAPPING = {
    "LABEL_0": "Negative",
    "LABEL_1": "Neutral",
    "LABEL_2": "Positive"
}

# 4. Definizione degli Endpoint

@app.get("/")
def home():
    """Endpoint di controllo per Hugging Face Spaces"""
    return {
        "message": "Il server RoBERTa Sentiment Analysis è attivo!",
        "status": "healthy",
        "model_used": MODEL_NAME,
        "docs_url": "/docs"
    }

@app.post("/predict")
def predict_sentiment(request: TextRequest):
    """Endpoint per l'analisi del sentiment"""
    if not sentiment_pipeline:
        raise HTTPException(status_code=500, detail="Il modello non è caricato correttamente sul server.")
    
    if not request.text.strip():
        raise HTTPException(status_code=400, detail="Il testo fornito non può essere vuoto.")
    
    try:
        # Eseguiamo l'inferenza
        result = sentiment_pipeline(request.text)[0]
        
        # Convertiamo l'etichetta del modello in qualcosa di leggibile (se presente nella mappa)
        raw_label = result["label"]
        friendly_label = LABEL_MAPPING.get(raw_label, raw_label)
        
        return {
            "text": request.text,
            "sentiment": friendly_label,
            "confidence_score": round(result["score"], 4),
            "raw_label": raw_label
        }
    except Exception as e:
        raise HTTPException(status_code=500, detail=f"Errore durante l'elaborazione: {str(e)}")

# 5. Avvio del server (obbligatorio su porta 7860 per Hugging Face)
if __name__ == "__main__":
    uvicorn.run(app, host="0.0.0.0", port=7860)
    

2.3 DockerFile per l'esecuzione del Docker Container che lancia l'app di Model Serving <br>

2.3) DockerFile
```text
FROM python:3.9

# Crea una directory di lavoro
WORKDIR /code

# Copia i requisiti e installali
COPY ./requirements.txt /code/requirements.txt
RUN pip install --no-cache-dir --upgrade -r /code/requirements.txt

# Configura un utente non-root per la sicurezza di Hugging Face
RUN useradd -m -u 1000 user
USER user
ENV HOME=/home/user \
	PATH=/home/user/.local/bin:$PATH

# Imposta la directory di lavoro dell'utente
WORKDIR $HOME/app

# Copia il resto del codice nell'applicazione
COPY --chown=user . $HOME/app

# Avvia l'applicazione FastAPI sulla porta 7860
CMD [" ????? ", "app.py"]
# Nota: puoi usare uvicorn direttamente se preferisci:
CMD ["uvicorn", "app:app", "--host", "0.0.0.0", "--port", "7860"]


**ANALISI MODEL SERVING** <br>

3. Caricamento dei file sullo Space di Hugging Face <br>
Puoi caricare questi tre file in due modi:
Via Git: Clona lo Space sul tuo PC usando il comando Git fornito nella pagina iniziale dello Space,
incolla i file, fai il commit e il push.
```text
cd C:\MyStudy\Corsi Prof AI\AIEngineering\Corso 10 MLops Machine Learning in Prod\ProjectMachineInnovators\app
# Init del progetto
git init
# Aggiungi tutti i files a git
git add .
# Fai la prima commit del progetto
git commit -m "Initial commit of app (model serving machine innovation)"
# collega il repository locale a quello remoto github oppure fastapi
git remote add origin https://github.com/francescofrigerio/profai_10_mlops_prj_machineinnovation.git

# rinomina il branch principale a main
git branch -M main
#  invia i tuoi commit locali al repository remoto su GitHub.
git push -u origin main




**ANALISI MODEL SERVING** <br>

4. Building Space di Hugging Face <br>
Appena caricherai i file (in particolare il Dockerfile), Hugging Face avvierà automaticamente il processo di Building.

Vedrai un badge in alto con scritto Building che diventerà Running se tutto è andato a buon fine.

Se esce Runtime Error, clicca su Logs per vedere cosa è andato storto nel codice o nelle dipendenze.

Una volta che è su Running, vedrai l'interfaccia web della tua app. Se hai configurato solo endpoint API,
vedrai una schermata bianca o il JSON del tuo endpoint /. Ti basterà aggiungere /docs
all'URL dello Space per accedere all'interfaccia interattiva di Swagger UI e testare la tua API!



**ANALISI MODEL SERVING** <br>
5. Test dell'applicazione <br>
Come testarlo una volta avviato: <br>
Una volta che lo Space è nello stato Running, vai nella scheda <br> dell'applicazione e aggiungi /docs alla fine dell'URL  <br>
(es: https://huggingface.co/spaces/tuo-username/tuo-space/docs). <br>

Vedrai l'interfaccia di Swagger. <br>

Clicca su POST /predict,  <br>
seleziona Try it out, inserisci una frase come <br>
 "Questo prodotto è fantastico, lo adoro!"  <br>
 nel corpo del JSON e clicca Execute.  <br>
 Riceverai una risposta simile a questa: <br>
{<br>
  "text": "Questo prodotto è fantastico, lo adoro!", <br>
  "sentiment": "Positive", <br>
  "confidence_score": 0.9854, <br>
  "raw_label": "LABEL_2" <br>
} <br>


**ANALISI DELLA SCHEDULAZIONE CON AIRFLOW** <br>

1. **Congigurazione environment**  <br>
Creare un nuovo codespace oppure modificare il Machine Type con 4 core  <br>
Creiiamo un nuovo branch dev_scheduling con il comando git checkout  <br>

git checkout -b dev-scheduling  <br>

creare la cartella scheduling  <br>

2. **Documentazione online**  <br>
Siamo pronti per usare AirFlow con docker-compose
cercare su google 'airflow docker compose'
troviamo la seguente pagina di apache airflow progetto open source  <br>

https://airflow.apache.org/docs/apache-airflow/stable/howto/docker-compose/index.html  <br>

Nella sezione 'Before you Begin' della pagina viene specificato come
serve avere installato docker e docker-compose  <br>
ma usando un github codespace siamo già a posto.  <br>


3. **Donwload docker-compose.yml**  <br>
Nella sezione fetcher docker-compose.yml si può copiare il comando curl
per scaricare dal terminale del code space il file docker-compose.yml
Si può anche accedere col browser all'url e fare copia e incolla dalla pagina.
cd scheduling # in modo che il file scaricato sia dove vogliamo noi
curl -LfO 'https://airflow.apache.org/docs/apache-airflow/3.2.1/docker-compose.yaml'


4. **Lista dei servizi** .  <br>
Nella pagina di air flow dedicata a docker-compose
subito dopo il comando per il download
abbiamo la lista dei servizi.  <br>


airflow-scheduler - The scheduler monitors all tasks and Dags, then triggers   <br>
the task instances once their dependencies are complete.   <br>

airflow-dag-processor - The Dag processor parses Dag files.  <br>

airflow-api-server - The api server is available at http://localhost:8080.  <br>

airflow-worker - The worker that executes the tasks given by the scheduler.  <br>

airflow-triggerer - The triggerer runs an event loop for deferrable tasks.  <br>

airflow-init - The initialization service.  <br>

postgres - The database.  <br>

redis - The redis - broker that forwards messages from scheduler to worker  <br>

5. **Lista dir sincronizzate tra il nostro pc e il container**  <br>

Dalla documentazione dopo il download: <br>
Some directories in the container are mounted, which means that their contents are synchronized
 between your computer and the container. <br>

./dags - you can put your Dag files here. <br>
./logs - contains logs from task execution and scheduler. <br>
./config - you can add custom log parser or add airflow_local_settings.py to  configure cluster policy. <br>
./plugins - you can put your custom plugins here. <br>

i dags (direct aciclic graph ) sono grafi diretti e aciclico
che rappresentano l'insieme dei passaggi con un inizio e una fine
che caratterizzano la pipeline <br>

logs dove sono salvati i logs <br>

config per configurazioni personalizzate <br>
plugin per eventuali plugin installati. <br>

copiarw le righe da eseguire nel file setup.sh per creare
le directory e generiamo un file .env con l'AIRFLOW ID
e le eseguiamo sul terminale <br>
cd scheduling <br>
mkdir -p ./dags ./logs ./plugins ./config <br>
echo -e "AIRFLOW_UID=$(id -u)" > .env <br>

Creo il file .gitignore sotto logs <br>
specificando di scartare tutto con * <br>
in modo da non salvare i logs sul repository <br>
eccetto il file .gitignore altrimenti diventa inutile il file. <br>

6. **Inizializzazione di airflow **  <br>
Proseguendo nella pagina di air flow
dedicata a docker compose c'è la sezione
dedicata all'inizializzazione dei database
che sono importanti in airflow perchè tengono
traccia delle esecuzioni tra uno step e l'altro.
Sono importanti perchè airflow andrà ad eseguire
in un ambiente distribuito in cloud su macchine diverse
e quindi se queste macchine vogliono parlarsi tra di loro
ci devono essere dei database che tengono traccia delle variabili
che le macchine sparse nel cloud vogliono che altre
macchine dopo di loro possono accedere .
E' molto importante avere delle entità terze come dei
database in produzione sempre funzionanti che storicizzano
le variabili in comune tra le diverse macchine.
Eseguiamo il seguente comando di iniznializzazione dei db
che scarica le immagini di redis(nosql) airflow e postgres(sql)

docker compose up airflow-init <br>

Alla fine dell'installazione viene creato
un utente di default con le seguenti credenziali <br>
login airflow  <br>
password airflow <br>

Se qualcosa dovesse andare storto
oppure volessimo ripulire la situazione
e ripartire da capo in modo pulito
i comandi seguenti non fanno
altro che cancellare tutta la dir dove
abbiamo posizionato il file docker-compose.yml
quindi nel nostro caso la dir scheduling . <br>

docker compose down --volumes --remove-orphans  <br>

// # Remove the entire directory where you downloaded docker-compose.yaml <br>
rm -rf '<DIRECTORY>' <br>

Per far partire il tutto una volta che tutto è a posto
biosgna eseguire <br>
docker compose up <br>

Prima di accedere con il browser
attendere che il servizio sia partito
effettivamente e modificare la visibilità
in pubblica della porta 8080 <br>

docker ps -> deve visualizzare healty e non unhealty <br>

https://scaling-parakeet-j7jxwwg944w3jwx-8080.app.github.dev/ <br>

se non si riesce ad accedere stoppare docker <br>
docker compose down <br>

editare il file docker-compose.yml
Trova la sezione environment: nel servizio airflow-webserver (e possibilmente in airflow-common se presente) e aggiungere o modificare queste tre voci: <br>

AIRFLOW__WEBSERVER__ENABLE_PROXY_FIX: 'True' <br>
AIRFLOW__WEBSERVER__EXPOSE_CONFIG: 'True' <br>
AIRFLOW__WEBSERVER__BASE_URL: https://scaling-parakeet-j7jxwwg944w3jwx-8080.app.github.dev <br>

ENABLE_PROXY_FIX: Dice ad Airflow di fidarsi degli header inviati dal proxy di GitHub (che agisce come un reverse proxy). <br>

BASE_URL: Deve essere esattamente l'URL postato(ttps://scaling-parakeet-j7jxwwg944w3jwx-8080.app.github.dev/ <br>), senza lo slash finale.
Questo istruisce Airflow su quale sia il suo "nome" pubblico ufficiale. <br>

7. **Applicare le modifiche** .
Dopo aver salvato il file, riavviare i container per caricare la nuova configurazione: <br>

docker-compose up -d <br>

accedere come user e pwd di default via browser ad airflow<br>

Sulla sx si vedono i dag  <br>
sulla dx il dettaglio del singolo dag <br>
Non abbiamo nulla attivo. <br>
E' possibile configurare l'orario del sistema
ed effettuare configurazioni ad esempio
per scatenare l'invio di email. <br>
Sono previsti circa 74 dag di default nessuno
dei quali è attivo e se dovessero partire
tutti insieme mandrebbero in crash la macchina.

**ARCHITETTURA DEL SISTEMA (CI CD TRAINING PIPELINE)**

                 ┌─────────────────────┐
                 │  TRAINING PIPELINE  │
                 │                     │
                 │ MLflow              │
                 │ SQLite              │
                 │ Grafana             │
                 └─────────┬───────────┘
                           │
                           │ produce model e grafici
                           ▼

                 ┌─────────────────────┐
                 │  GITHUB REPOSITORY  │
                 │                     │
                 │ model           │
                 │ FastAPI app         │
                 │ Dockerfile          │
                 └─────────┬───────────┘
                           │
                           │ deploy
                           ▼

                 ┌─────────────────────┐
                 │ HUGGING FACE SPACE  │
                 │                     │
                 │ FastAPI             │
                 │ Loaded Model        │
                 │ /predict API        │
                 └─────────────────────┘

                        ┌────────────────────┐
                        │     GitHub Repo    │
                        │                    │
                        │  - source code     │
                        │  - docs            │
                        │  - Dockerfile      │
                        │  - requirements    │
                        └─────────┬──────────┘
                                  │
                                  │ Git push / CI-CD
                                  ▼


══════════════════════════════════════════════════════════════

        ┌──────────────────────────────┐
        │        APACHE AIRFLOW        │
        │                              │
        │ DAG:                         │
        │ - data ingestion             │
        │ - preprocessing              │
        │ - training                   │
        │ - testing                    │
        │ - monitoring                 │
        │ - retraining                 │
        └─────────────┬────────────────┘
                      │
                      ▼

        ┌──────────────────────────────┐
        │           MLFLOW             │
        │                              │
        │ - experiments                │
        │ - metrics                    │
        │ - params                     │
        │ - artifacts                  │
        │ - model registry             │
        └─────────────┬────────────────┘
                      │
                      ▼

        ┌──────────────────────────────┐
        │         SQLITE DB            │
        │                              │
        │ model_metrics table          │
        │ prediction_logs table        │
        │ metrics table          │
        └─────────────┬────────────────┘
                      │
                      ▼

        ┌──────────────────────────────┐
        │           GRAFANA            │
        │                              │
        │ Dashboards:                  │
        │ - accuracy                   │
        │ - latency                    │
        │ - drift                      │
        │ - prediction stats           │
        └──────────────────────────────┘

#**SISTEMA FINALE** <br>
```mermaid
 ┌─────────────────────────────────────────────────────────────────────────┐
 │                            APACHE AIRFLOW                               │
 │                                                                         │
 │   ┌──────────────────────┐             ┌────────────────────────────┐   │
 │   │  DAG: RETRAINING     │             │  DAG: DAILY MONITORING     │   │
 │   │  - "mode":"prod"     |             │  - "mode":"prod"           │   │
 │   │  - Mensile ordinario │             │  - Gira OGNI GIORNO        │   │
 │   │  - O su emergenza ◄──┼─────────────┼─ - Controlla soglia drift  │   │
 │   └──────────┬───────────┘ (Trigger    └─────────────┬──────────────┘   │
 └──────────────┼────────────── if drift)               │                  │
                │                                       │                  │
    Chiamata API ("mode":"prod"/"demo")      Chiamata API (Estrae JSON)    │
                │                                       │                  │
                ▼                                       ▼                  │
 ┌─────────────────────────────────────────────────────────────────────────┐
 │                           GITHUB REPOSITORY                             │
 │                                                                         │
 │   [WORKFLOW: RETRAIN] ──────────────────► [WORKFLOW: MONITORING]        │
 │   mode:"prod"                                                           |
 |   Addestra il modello (2 epoche)          Aggiorna grafici e            │
 │   e fa il push del nuovo .db              genera il file JSON metriche  │
 |                                                                         |
 |   mode:"demo"                                                           |
 |   Addestra il modello (meno di 1 epoca)   Aggiorna grafici e            │
 │   e fa il push del nuovo .db              genera il file JSON metriche  │
 └──────────────┬───────────────────────────────────────▲──────────────────┘
                │                                       │
                │ Deploy automatico                     │ Legge dati reali
                ▼                                       │
 ┌──────────────────────────┐             ┌─────────────┴──────────────┐
 │   HUGGING FACE SPACE     │             │         SQLITE DB          │
 │                          │             │                            │
 │   Servizio FastAPI       │────────────►│   Tabelle con predizioni   │
 │   in produzione          │ (Log di     │   e performance reali      │
 └──────────────────────────┘  output)    └────────────────────────────┘

 ```
